```python
# Define target variable
TARGET = "covid_vaccine"

# Save testing IDs
test_ids = test["respondent_id"].copy()

# Separate training features
X = train.drop(
    columns=["respondent_id", TARGET]
)

# Separate target
y = train[TARGET].astype(int)

# Separate testing features
X_test = test.drop(
    columns=["respondent_id"]
)

print("Training features:", X.shape)
print("Training target:", y.shape)
print("Testing features:", X_test.shape)
```

```python
# Identify categorical columns
categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Identify numerical columns
numerical_cols = [
    column for column in X.columns
    if column not in categorical_cols
]

print("Numerical columns:")
print(numerical_cols)

print("\nCategorical columns:")
print(categorical_cols)

print("\nTotal numerical columns:",
      len(numerical_cols))

print("Total categorical columns:",
      len(categorical_cols))
```

```python
# Missing values in training features

missing_summary = pd.DataFrame({

    "Training Missing": X.isnull().sum(),

    "Testing Missing": X_test.isnull().sum()

})

missing_summary["Training Missing %"] = (
    X.isnull().mean() * 100
)

missing_summary["Testing Missing %"] = (
    X_test.isnull().mean() * 100
)

display(
    missing_summary.sort_values(
        by="Training Missing %",
        ascending=False
    )
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[

    ("imputer",
     SimpleImputer(strategy="median")),

    ("scaler",
     StandardScaler())

])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[

    ("imputer",
     SimpleImputer(strategy="most_frequent")),

    ("encoder",
     OneHotEncoder(
         handle_unknown="ignore",
         sparse_output=False
     ))

])

# Combine both pipelines
preprocessor = ColumnTransformer(

    transformers=[

        ("numerical",
         numeric_transformer,
         numerical_cols),

        ("categorical",
         categorical_transformer,
         categorical_cols)

    ],

    remainder="drop"

)

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_valid, y_train, y_valid = train_test_split(

    X,

    y,

    test_size=0.25,

    random_state=42,

    stratify=y

)

print("Training records:", X_train.shape[0])

print("Validation records:", X_valid.shape[0])

print("\nTraining acceptance rate:",
      round(y_train.mean() * 100, 2), "%")

print("Validation acceptance rate:",
      round(y_valid.mean() * 100, 2), "%")
```

```python
# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation and testing data
X_valid_processed = preprocessor.transform(X_valid)

X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:",
      X_train_processed.shape)

print("Processed validation shape:",
      X_valid_processed.shape)

print("Processed testing shape:",
      X_test_processed.shape)

print("\nMissing values after preprocessing:",
      np.isnan(X_train_processed).sum())

print("\nPreprocessing completed successfully!")
```

```python
from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    HistGradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    log_loss,
    brier_score_loss,
    confusion_matrix,
    classification_report
)

print("Machine learning algorithms imported successfully!")
```

```python
models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000,
            C=0.5,
            class_weight="balanced",
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=400,
            max_depth=None,
            min_samples_split=5,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

    "Extra Trees":
        ExtraTreesClassifier(
            n_estimators=400,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

    "HistGradientBoosting":
        HistGradientBoostingClassifier(
            max_iter=200,
            learning_rate=0.05,
            l2_regularization=2,
            random_state=42
        )
}

print("Five scikit-learn models defined successfully!")
```

```python
trained_models = {}

for name, algorithm in models.items():

    print("\nTraining:", name)

    model_pipeline = Pipeline(steps=[

        ("preprocessor", preprocessor),

        ("classifier", algorithm)

    ])

    model_pipeline.fit(
        X_train,
        y_train
    )

    trained_models[name] = model_pipeline

    print(name, "trained successfully!")

print("\nAll five models have completed training!")
```

```python
validation_probabilities = {}
validation_predictions = {}

for name, model in trained_models.items():

    # Predict probability of vaccine acceptance
    probabilities = model.predict_proba(X_valid)[:, 1]

    # Use 0.50 as the initial classification threshold
    predictions = (probabilities >= 0.50).astype(int)

    validation_probabilities[name] = probabilities
    validation_predictions[name] = predictions

    print("\nModel:", name)

    print("ROC-AUC:",
          round(roc_auc_score(y_valid, probabilities), 4))

    print("Accuracy:",
          round(accuracy_score(y_valid, predictions), 4))
```

```python
results = []

for name in trained_models:

    probabilities = validation_probabilities[name]
    predictions = validation_predictions[name]

    results.append({

        "Model": name,

        "Accuracy": accuracy_score(
            y_valid, predictions
        ),

        "Precision": precision_score(
            y_valid, predictions, zero_division=0
        ),

        "Recall": recall_score(
            y_valid, predictions, zero_division=0
        ),

        "F1 Score": f1_score(
            y_valid, predictions, zero_division=0
        ),

        "ROC-AUC": roc_auc_score(
            y_valid, probabilities
        ),

        "PR-AUC": average_precision_score(
            y_valid, probabilities
        ),

        "Log Loss": log_loss(
            y_valid, probabilities
        ),

        "Brier Score": brier_score_loss(
            y_valid, probabilities
        )

    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="ROC-AUC",
    ascending=False
)

display(
    results_df.round(4)
)
```

```python
# ==========================================
# STEP 17: COST ESTIMATION AND OPTIMISATION
# CELL 43: DEFINE COST ASSUMPTIONS
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import shutil

from sklearn.base import clone
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Illustrative cost assumptions in USD
VACCINE_COST = 57.45
HOSPITAL_COST = 11275.00

# Illustrative vaccine effectiveness against hospitalisation
VACCINE_EFFECTIVENESS = 0.86

# Assignment instruction: assume 100% probability of infection
INFECTION_PROBABILITY = 1.00

# Illustrative baseline hospitalisation risks by age group
AGE_HOSPITAL_RISK = {
    "18 - 34 Years": 0.019,
    "35 - 44 Years": 0.028,
    "45 - 54 Years": 0.047,
    "55 - 64 Years": 0.0825,
    "65+ Years": 0.18
}

# Illustrative relative increase per chronic condition
CHRONIC_CONDITION_RISK_INCREASE = 0.25

print("Cost assumptions loaded successfully!")
print("Vaccine cost: $", VACCINE_COST)
print("Hospitalisation cost: $", HOSPITAL_COST)
print("Vaccine effectiveness:", VACCINE_EFFECTIVENESS)
```

```python
# ==========================================
# CELL 44: ESTIMATE HOSPITALISATION RISK
# ==========================================

def estimate_unvaccinated_hospital_risk(data):

    age_risk = (
        data["age_group"]
        .fillna("45 - 54 Years")
        .map(AGE_HOSPITAL_RISK)
        .fillna(0.047)
        .to_numpy(dtype=float)
    )

    chronic_count = (
        pd.to_numeric(
            data["number_chronic_med_conditions"],
            errors="coerce"
        )
        .fillna(0)
        .clip(lower=0)
        .to_numpy(dtype=float)
    )

    risk = age_risk * (
        1 + CHRONIC_CONDITION_RISK_INCREASE * chronic_count
    )

    return np.clip(
        risk * INFECTION_PROBABILITY,
        0,
        0.95
    )


def calculate_expected_costs(probability_acceptance, data):

    risk_unvaccinated = estimate_unvaccinated_hospital_risk(data)

    risk_vaccinated = (
        risk_unvaccinated * (1 - VACCINE_EFFECTIVENESS)
    )

    # Decision 1: Send vaccine
    cost_send = (
        VACCINE_COST
        +
        (
            probability_acceptance * risk_vaccinated
            +
            (1 - probability_acceptance) * risk_unvaccinated
        ) * HOSPITAL_COST
    )

    # Decision 2: Do not send vaccine.
    # Pay compensation to people who would have accepted it.
    cost_not_send = (
        probability_acceptance * VACCINE_COST
        +
        risk_unvaccinated * HOSPITAL_COST
    )

    return cost_send, cost_not_send


# Check hospitalisation risk on the validation data
validation_hospital_risk = estimate_unvaccinated_hospital_risk(X_valid)

print("Hospitalisation risk calculated!")
print("Average estimated hospitalisation risk:",
      round(validation_hospital_risk.mean(), 4))

print("Minimum risk:", round(validation_hospital_risk.min(), 4))
print("Maximum risk:", round(validation_hospital_risk.max(), 4))
```

```python
# ==========================================
# CELL 45: COST-BASED MODEL OPTIMISATION
# ==========================================

cost_models = trained_models.copy()

cost_models["Tuned Random Forest"] = random_search.best_estimator_

cost_results = []
chosen_thresholds = {}

for model_name, model in cost_models.items():

    probabilities = model.predict_proba(X_valid)[:, 1]

    send_cost, no_send_cost = calculate_expected_costs(
        probabilities,
        X_valid
    )

    best_threshold = None
    lowest_cost = float("inf")

    # Test thresholds between 0.01 and 0.99
    for threshold in np.arange(0.01, 1.00, 0.01):

        send_vaccine = probabilities >= threshold

        total_cost = np.where(
            send_vaccine,
            send_cost,
            no_send_cost
        ).sum()

        if total_cost < lowest_cost:

            lowest_cost = total_cost
            best_threshold = round(float(threshold), 2)

    chosen_thresholds[model_name] = best_threshold

    predictions = (
        probabilities >= best_threshold
    ).astype(int)

    cost_results.append({

        "Model": model_name,

        "Best Threshold": best_threshold,

        "Expected Total Cost": lowest_cost,

        "Cost Per Person": lowest_cost / len(X_valid),

        "ROC-AUC": roc_auc_score(y_valid, probabilities),

        "Accuracy": accuracy_score(y_valid, predictions),

        "Precision": precision_score(
            y_valid, predictions, zero_division=0
        ),

        "Recall": recall_score(
            y_valid, predictions, zero_division=0
        ),

        "F1 Score": f1_score(
            y_valid, predictions, zero_division=0
        )
    })


cost_results_df = pd.DataFrame(cost_results)

cost_results_df = cost_results_df.sort_values(
    by="Expected Total Cost",
    ascending=True
).reset_index(drop=True)

print("COST-OPTIMISED MODEL COMPARISON")

display(cost_results_df.round(4))

best_model_name = cost_results_df.iloc[0]["Model"]

print("\nLowest estimated validation cost model:")
print(best_model_name)

print(
    "Optimised threshold:",
    chosen_thresholds[best_model_name]
)
```
